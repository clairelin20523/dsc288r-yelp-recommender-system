# Data Exploration

## Helper methods

```python
def print_num_cols(df):
    """
    Get number of columns
    """
    num_cols = len(df.columns)
    print('There is a total of %d columns' % num_cols)
```

```python
def print_num_missing(df):
    """
    Get number of missing values for each column
    """
    columns = df.columns
    num_cols = len(columns)
    for i in range(num_cols):
        missing = df[columns[i]].isnull().sum()
        print("'{}' column has {} missing values".format(columns[i], missing))
```

```python
def print_types(df):
    """
    """
    data_types = df.dtypes
    columns = df.columns
    num_cols = len(columns)
    for i in range(num_cols):
        print("'%s' column is of type '%s'" % (columns[i], data_types.iloc[i]))
```

## Flatten categories

```python
review_subset = df_review_data[['business_id', 'stars']]
business_subset = df_business_data[['business_id', 'categories']]
business_subset2 = df_business_data[['business_id', 'name']]


del df_review_data, df_business_data



review_counts = review_subset.groupby('business_id').size().reset_index(name='review_count')

merged_df = pd.merge(review_subset, business_subset, on='business_id', how='left')

merged_df = merged_df.dropna(subset=['categories'])

category_list = [(category, stars) 
                 for categories, stars in zip(merged_df['categories'].str.split(', '), merged_df['stars']) 
                 for category in categories]


category_df = pd.DataFrame(category_list, columns=['category', 'stars'])

del merged_df
category_ratings = category_df.groupby('category')['stars'].mean().reset_index()

category_ratings = category_ratings.sort_values(by='stars', ascending=False)

print(category_ratings.head(20))
category_ratings.head(20).to_csv('category_ratings_top20.csv', index=False)
```

```python
top_categories = category_ratings[category_ratings['stars'] == 5.0]

top_category_df = category_df[category_df['category'].isin(top_categories['category'])]

top_category_count = top_category_df.groupby('category')['stars'].count()
top_category_count = top_category_count.reset_index(name = 'business_counts')

top_category_table = pd.merge(top_categories, top_category_count, on = 'category', how = 'left')

print(top_category_table)

top_category_table.to_csv('top_categories.csv', index = False)
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


review_counts = review_subset.groupby('business_id').size().reset_index(name='review_count')


review_counts['z_score'] = (review_counts['review_count'] - review_counts['review_count'].mean()) / review_counts['review_count'].std()


anomalies = review_counts[review_counts['z_score'].abs() > 3]


plt.figure(figsize=(10, 5))
sns.histplot(review_counts['review_count'], bins=50, kde=True, color='skyblue')
plt.axvline(anomalies['review_count'].min(), color='red', linestyle='dashed', label="Anomaly Threshold")
plt.axvline(anomalies['review_count'].max(), color='red', linestyle='dashed')
plt.xlabel('Review Count', fontsize = 12)
plt.ylabel('Frequency', fontsize = 12)
plt.title('Distribution of Review Counts with Anomalies', fontsize = 18)
plt.legend()
plt.savefig('anomoly.png', dpi=300)
plt.show()


print(anomalies.sort_values(by='review_count', ascending=False))
```

```python
file_path = 'yelp_academic_dataset_review.json'

chunk_size = 100000
chunks = []
for chunk in pd.read_json(file_path, lines=True, chunksize=chunk_size):
    chunks.append(chunk)

df_review_data = pd.concat(chunks, ignore_index=True)

print(df_review_data.head())
      
df_review_data['date'] = pd.to_datetime(df_review_data['date'])
review_counts_time = df_review_data.groupby(['business_id', df_review_data['date'].dt.to_period("M")])['review_id'].count().reset_index()
review_counts_time.columns = ['business_id', 'month', 'review_count']
review_counts_time['z_score'] = review_counts_time.groupby('business_id')['review_count'].transform(lambda x: (x - x.mean()) / x.std())
review_spikes = review_counts_time[review_counts_time['z_score'] > 3]

print(review_spikes.sort_values(by='z_score', ascending=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

review_counts_time['month'] = review_counts_time['month'].astype(str)
review_counts_time['month'] = pd.to_datetime(review_counts_time['month'])

monthly_total_reviews = review_counts_time.groupby('month')['review_count'].sum()

monthly_spike_reviews = review_spikes.groupby('month')['review_count'].sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly_total_reviews.index, monthly_total_reviews.values, marker='o', linestyle='-', label="Total Reviews")
plt.plot(monthly_spike_reviews.index, monthly_spike_reviews.values, marker='o', linestyle='-', label="Outlier Reviews (Spikes)", color='red')

plt.xlabel("Month")
plt.ylabel("Review Count")
plt.title("Total Reviews vs Outlier Reviews (2019-2021)")
plt.xticks(rotation=45)
plt.legend()
plt.grid()

# Show plot
plt.show()
```
